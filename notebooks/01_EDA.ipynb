```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.db.connection import get_connection

plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

```python
engine = get_connection()

movies = pd.read_sql("SELECT * FROM movies", engine)
ratings = pd.read_sql("SELECT * FROM ratings", engine)
tags = pd.read_sql("SELECT * FROM tags", engine)

print(f"Movies: {movies.shape}")
print(f"Ratings: {ratings.shape}")
print(f"Tags: {tags.shape}")
```

```python
print(movies.dtypes)
print(ratings.dtypes)
print(movies.isnull().sum())
print(ratings.isnull().sum())
```

```python
plt.figure(figsize=(10, 5))
ratings['rating'].value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
ratings_per_movie = ratings.groupby('movie_id')['rating'].count()

plt.figure(figsize=(10, 5))
plt.hist(ratings_per_movie, bins=50, color='steelblue', edgecolor='black')
plt.title('Ratings Per Movie')
plt.xlabel('Number of Ratings')
plt.ylabel('Number of Movies')
plt.yscale('log')
plt.tight_layout()
plt.show()
```
